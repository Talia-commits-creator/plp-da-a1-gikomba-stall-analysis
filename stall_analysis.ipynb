This project analyses 14 days of sales and stock data for Wanjiku's fresh-produce stall at Gikomba Market.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Daily Sales Analysis

I used Python's built-in functions to calculate total, average, highest and lowest sales. I also identified the best and worst days using their positions in the sales list.



```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

print("Total sales:", total_sales)
print("Average daily sales:", round(average_sales, 2))
print("Highest sales:", highest_sales)
print("Lowest sales:", lowest_sales)

print("Best day:", days[best_index],
      "Week:", best_index // 7 + 1)

print("Worst day:", days[worst_index],
      "Week:", worst_index // 7 + 1)
```

## Task 2: Week-on-Week Growth

I created a reusable function using list slicing to calculate the sales totals for each week. I then calculated the percentage growth from Week 1 to Week 2.

```python
def week_totals(sales):
    """Return the total sales for Week 1 and Week 2."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


week1, week2 = week_totals(daily_sales)

growth = ((week2 - week1) / week1) * 100

print("Week 1 sales:", week1)
print("Week 2 sales:", week2)
print("Week-on-week growth:", round(growth, 2), "%")
```

## Task 3: Daily Sales Classification

I created a function to classify each day as Strong, Normal or Weak based on its sales compared with the overall average. I used a dictionary to count the days in each category.

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal or Weak."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    counts[category] += 1

print("Daily Sales Categories:")
for category, count in counts.items():
    print(category, ":", count)
```

## Task 4: Weekday vs Weekend Performance

I calculated the average sales for weekdays (Monday to Friday) and weekends (Saturday and Sunday). I then compared weekend performance with weekday performance as a percentage.

```python
weekdays = daily_sales[0:5] + daily_sales[7:12]
weekends = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekdays) / len(weekdays)
weekend_average = sum(weekends) / len(weekends)

difference = ((weekend_average - weekday_average)
              / weekday_average) * 100

print("Weekday average:", round(weekday_average, 2))
print("Weekend average:", round(weekend_average, 2))
print("Weekend increase:", round(difference, 2), "%")
```

## Task 5: Stock Reordering and Gross Margins

I used the stock dictionary to identify products whose quantities are at or below their reorder levels. I also calculated the gross margin percentage for each product.

```python
def products_to_reorder(stock):
    """Return products at or below their reorder levels."""
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_list = products_to_reorder(stock)

print("Products to reorder:", reorder_list)

print("\nGross Margins:")
for product, details in stock.items():
    margin = ((details["price"] - details["cost"])
              / details["price"]) * 100

    print(product, ":", round(margin, 2), "%")
```

## Task 6: Stall Performance Report

I combined the results from the previous tasks into a formatted business report using f-strings and thousands separators for Kenyan shilling amounts.

```python
print("=" * 45)
print("       STALL PERFORMANCE REPORT")
print("=" * 45)

print(f"Total Sales:          KES {total_sales:,.2f}")
print(f"Average Daily Sales:  KES {average_sales:,.2f}")
print(f"Highest Sales:        KES {highest_sales:,.2f}")
print(f"Lowest Sales:         KES {lowest_sales:,.2f}")

print("-" * 45)
print(f"Week 1 Sales:         KES {week1:,.2f}")
print(f"Week 2 Sales:         KES {week2:,.2f}")
print(f"Sales Growth:         {growth:.2f}%")

print("-" * 45)
print(f"Weekday Average:      KES {weekday_average:,.2f}")
print(f"Weekend Average:      KES {weekend_average:,.2f}")

print("-" * 45)
print("Sales Classification:")

for category, count in counts.items():
    print(f"{category:<12}: {count} days")

print("-" * 45)
print("Products to Reorder:")

for product in reorder_list:
    print(f"- {product}")

print("=" * 45)
print("             END OF REPORT")
print("=" * 45)
```
